## 1. Imports

```python
import pathlib, random
from langchain_core.documents import Document
print("Imports OK")
```

## 2. Load Knowledge Base from Files


```python
KB_PATH = pathlib.Path("knowledge-base")

def load_knowledge_base(kb_path: pathlib.Path) -> list:
    """
    Walk every .md file in kb_path and return a list of LangChain Documents.
    Each Document carries source (relative path) and category (subfolder) metadata.
    """
    if not kb_path.exists():
        raise FileNotFoundError(
            f"knowledge-base folder not found at '{kb_path.resolve()}'\n"
            "Make sure you are running this notebook from the project root folder."
        )

    docs = []
    for md_file in sorted(kb_path.rglob("*.md")):
        relative = md_file.relative_to(kb_path)   # e.g. products/nova_phone_x1.md
        category = relative.parts[0]              # e.g. 'products'
        text     = md_file.read_text(encoding="utf-8")
        docs.append(Document(
            page_content=text,
            metadata={
                "source"  : str(relative),
                "category": category,
            }
        ))
    return docs


raw_docs = load_knowledge_base(KB_PATH)
print(f"Loaded {len(raw_docs)} documents from knowledge-base/\n")

# Show every loaded document
for doc in raw_docs:
    print(f"  [{doc.metadata['category']:<20}]  {doc.metadata['source']}")
```

## 3. Sanity Checks

```python
from collections import Counter

# Count by category
cats = Counter(d.metadata['category'] for d in raw_docs)
print('Documents per category:')
for cat, n in sorted(cats.items()):
    print(f'  {cat:<22} {n} docs')

print(f'\nTotal: {len(raw_docs)} documents')

# Show 3 random document metadata + preview
print('\nRandom sample (3 docs):')
for doc in random.sample(raw_docs, 3):
    print(f"\n  source   : {doc.metadata['source']}")
    print(f"  category : {doc.metadata['category']}")
    print(f"  length   : {len(doc.page_content)} chars")
    print(f"  preview  : {doc.page_content[:100].strip()!r}")
```
